```python
import pandas as pd


df = pd.read_csv('n_movies.csv')

df['year'] = df['year'].astype(str).str.extract(r'(\d+)')

df['year'] = pd.to_numeric(df['year'], errors='coerce')

film__baru = df[df['year'] > 2000]

film__baru.head()

film__baru.to_csv('film_baru.csv', index=False)
```
